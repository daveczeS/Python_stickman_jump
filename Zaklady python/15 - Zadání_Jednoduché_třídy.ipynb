````python
class Clovek:
    def __init__(self, jmeno: str, prijmeni: str, telefon: int,
                 email: str, adresa: str, pojistovna: str):

        self.__jmeno = None
        self.prijmeni = None
        self.telefon = None
        self.email = None
        self.adresa = None
        self.pojistovna = None

        self.set_jmeno(jmeno)
        self.set_prijmeni(prijmeni)
        self.set_telefon(telefon)
        self.set_email(email)
        self.set_adresa(adresa)
        self.set_pojistovna(pojistovna)


    def get_jmeno(self):
        return self.__jmeno

    def set_jmeno(self, nove_jmeno):
        if type(nove_jmeno) == str and nove_jmeno.strip():
            self.__jmeno = nove_jmeno
        else:
            print("Chyba: jméno musí být neprázdný text")

    def get_prijmeni(self):
        return self.prijmeni

    def set_prijmeni(self, nove_prijmeni):
        if type(nove_prijmeni) == str and nove_prijmeni.strip():
            self.prijmeni = nove_prijmeni
        else:
            print("Chyba: příjmení musí být neprázdný text")


    def get_telefon(self):
        return self.telefon

    def set_telefon(self, novy_telefon):
        if type(novy_telefon) == int and len(str(novy_telefon)) <= 9:
            self.telefon = novy_telefon
        else:
            print("Chyba: telefon musí být číslo s maximálně 9 číslicemi")

  
    def get_email(self):
        return self.email

    def set_email(self, novy_email):
        if type(novy_email) == str and "@" in novy_email:
            self.email = novy_email
        else:
            print("Chyba: email musí obsahovat znak @")

   
    def get_adresa(self):
        return self.adresa

    def set_adresa(self, nova_adresa):
        if type(nova_adresa) == str and nova_adresa.strip():
            self.adresa = nova_adresa
        else:
            print("Chyba: adresa nesmí být prázdná")


    def get_pojistovna(self):
        return self.pojistovna

    def set_pojistovna(self, nova_pojistovna):
        if type(nova_pojistovna) == str and nova_pojistovna.strip():
            self.pojistovna = nova_pojistovna
        else:
            print("Chyba: pojišťovna nesmí být prázdná")

    def __str__(self):
        return f"""
jmeno: {self.__jmeno}
prijmeni: {self.prijmeni}
telefon: {self.telefon}
email: {self.email}
adresa: {self.adresa}
pojistovna: {self.pojistovna}
"""
    


clovek1= Clovek("Dan","novotny",7730255239,"ahoj.nevim@gmail.com","U parku 174","VZP")
clovek2= Clovek("David","nevim",783581963,"cau.idkgmail.com","shrekova bazina 13","VZP")
print(clovek1)
print(clovek2)
````
