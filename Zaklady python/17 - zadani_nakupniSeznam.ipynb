```python
# Pokud máme seznam a chceme vypsat jeho obsah pomocí metody __str__,
# musíme vytvořit metodu, která nám seznam převede na string
class Seznam:
    def __init__(self,seznam:list):
        self.seznam = seznam #vše co je v košíku
    def __str__(self):
        #return self.seznam # nevrací string, ale LIST
        return "{}".format(self.seznam) # to už projde
        #ale co když je vstupem seznamu jiná třída

sez = Seznam(["jablko","hrušky"])
print(sez)

class Polozka:
    def __init__(self,pocet:int,cena:float):
        self.cena = cena
        self.pocet = pocet
    def __str__(self):
        return "cena: {} a počet: {}".format(self.cena,self.pocet)

jabka = Polozka(5,3)
pomelo = Polozka(5,10)
sez = Seznam([jabka,pomelo])
print(sez) # print nám vypíše pouze třídu, je třeba dopsat metodu seznamToString()
#[<__main__.Polozka object at 0x785f7118dc00>, <__main__.Polozka object at 0x785f7118dde0>]

class Seznam:
    def __init__(self,seznam:list):
        self.seznam = seznam #vše co je v košíku
    def seznamToString(self): # metoda, která přepíše seznam do stringu
        text = ""
        for polozka in self.seznam:
            text += "cena: {} a počet: {}\n".format(polozka.cena,polozka.pocet)
        return text
    def __str__(self):
        # musím volat metodu seznamToString(),
        # aby mi vypsala jen string všeho v poli
        return "{}".format(self.seznamToString())
        #return "{}".format(self.seznam) # to nefunguje pro vstup Class

sez = Seznam([jabka,pomelo])
print(sez) # print nám vypíše pouze třídu, je třeba dopsat metodu seznamToString()
```

#2
Vytvořte třídu Seznam, která bude mít tyto atributy:

*   Název seznamu
*   Seznam položek
*   Maximální útrata (kolik mám u sebe peněz)

Ta bude obsahovat metody pro get, set, ToString a metody pro přidání a odebrání položky seznamu a zjištění ceny celého nákupu.


```python
class Polozka:
    def __init__(self, nazev, cena):
        self.nazev = nazev
        self.cena = cena

    def __str__(self):
        return f"{self.nazev} - {self.cena} Kč"


class Seznam:
    def __init__(self, nazev_seznamu, maximalni_utrata):
        self._nazev_seznamu = nazev_seznamu
        self._polozky = []
        self._maximalni_utrata = maximalni_utrata

    # přidání položky
    def pridat_polozku(self, polozka):
        self._polozky.append(polozka)

    # odebrání položky
    def odebrat_polozku(self, polozka):
        self._polozky.remove(polozka)

    # cena celého nákupu
    def cena_celkem(self):
        return sum(p.cena for p in self._polozky)

    # ToString
    def __str__(self):
        vystup = f"Seznam: {self._nazev_seznamu}\n"
        vystup += "Položky:\n"
        for p in self._polozky:
            vystup += f" - {p}\n"
        vystup += f"Celková cena: {self.cena_celkem()} Kč\n"
        vystup += f"Maximální útrata: {self._maximalni_utrata} Kč"
        return vystup
nakup = Seznam("Nákup do obchodu", 500)
nakup.pridat_polozku(Polozka("Chleba", 35))
nakup.pridat_polozku(Polozka("Mléko", 25))
nakup.pridat_polozku(Polozka("Sýr", 80))

print(nakup)
```

Přepište třídu Polozka tak, aby využíval Dictionary místo atributů i s metodama.

```python
class Polozka:
    def __init__(self, nazev, cena):
        self.data = {
            "nazev": nazev,
            "cena": cena
        }

    def get_nazev(self):
        return self.data["nazev"]

    def get_cena(self):
        return self.data["cena"]

    def __str__(self):
        return f"{self.data['nazev']} - {self.data['cena']} Kč"


class Seznam:
    def __init__(self, nazev_seznamu, maximalni_utrata):
        self._nazev_seznamu = nazev_seznamu
        self._polozky = []
        self._maximalni_utrata = maximalni_utrata

    # přidání položky
    def pridat_polozku(self, polozka):
        self._polozky.append(polozka)

    # odebrání položky
    def odebrat_polozku(self, polozka):
        self._polozky.remove(polozka)

    # cena celého nákupu
    def cena_celkem(self):
        return sum(p.get_cena() for p in self._polozky)

    # ToString
    def __str__(self):
        vystup = f"Seznam: {self._nazev_seznamu}\n"
        vystup += "Položky:\n"
        for p in self._polozky:
            vystup += f" - {p}\n"
        vystup += f"Celková cena: {self.cena_celkem()} Kč\n"
        vystup += f"Maximální útrata: {self._maximalni_utrata} Kč"
        return vystup

nakup = Seznam("Nákup do obchodu", 500)
nakup.pridat_polozku(Polozka("Chleba", 35))
nakup.pridat_polozku(Polozka("Mléko", 25))
nakup.pridat_polozku(Polozka("Sýr", 80))

print(nakup)
```
